```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 20, 25, 30]

# Plot vertical lines
for i in range(len(x)):
    plt.plot([x[i], x[i]], [0, y[i]], color='blue')

# Plot markers at the top
plt.plot(x, y, marker='o', linestyle='')

plt.title("Bar Chart Simulation using plt.plot()")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

```python
import matplotlib.pyplot as plt

# Dataset
x = [-3, -2, -1, 0, 1, 2, 3]
y = [-9, -4, -1, 0, 1, 4, 9]

# 1. Create a color logic list
# 'green' if positive, 'red' if negative, 'black' for zero
colors = ['green' if val > 0 else 'red' if val < 0 else 'black' for val in y]

plt.figure(figsize=(8, 5))

# 2. Plot using the color list
plt.bar(x, y, color=colors, edgecolor='black')

# 3. Add a baseline at y=0 to emphasize the divide
plt.axhline(0, color='black', linewidth=1)

plt.title('Conditional Coloring: Positive (Green) vs Negative (Red)')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Dataset
x = [1, 2, 3, 4, 5]
y = [3, 6, 2, 8, 7]

plt.figure(figsize=(8, 5))

# 1. Plot the main top line to define the "boundary"
plt.plot(x, y, color='skyblue', linewidth=3, marker='o')

# 2. Use a loop to create vertical lines from baseline (0) to y
# This simulates the "filled" look of an area chart
for i in range(len(x)):
    # plt.plot([x_start, x_end], [y_start, y_end])
    plt.plot([x[i], x[i]], [0, y[i]], color='skyblue', linewidth=2, alpha=0.6)

# 3. Add a baseline for visual stability
plt.axhline(0, color='black', linewidth=1)

plt.title("Simulated Area Chart (Vertical Line Method)")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```
