```python
import pandas as pd
weights = [55,60,62,65,70,72,75,80,85,90,95,100]
plt.hist(weights,bins= [50,60,70,80,90,100],color='green',alpha=0.6,density=True)
s=pd.Series(weights)
s.plot.density(bw_method='scott', color='blue', linestyle='--', linewidth=2)
plt.show()
```

```python
weights = [55,60,62,65,70,72,75,80,85,90,95,100]

fig, ax = plt.subplots(figsize=(8,5))

# Histogram with density scaling
ax.hist(weights, bins=[50,60,70,80,90,100], color='green', alpha=0.6,density=True)

# Convert list to Series
s = pd.Series(weights)

# Plot density on the same axes
s.plot.density(bw_method='scott', color='blue', linestyle='-', linewidth=2, ax=ax)

ax.set_title("Weights Histogram with Density Curve")
ax.set_xlabel("Weights")
ax.set_ylabel("Density")

plt.show()
```
