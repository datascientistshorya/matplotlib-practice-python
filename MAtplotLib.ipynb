## Creating different types of graphs

### 1. Line Graph
#### are commenl used to visualize treands over time or relationships between variables. We'll learn how to create visually appealing line graphs to represent such data.

```python
import matplotlib.pyplot as plt
x=[3,1,3,6,7,8]
y=[3,2,1,5,3,7]
plt.plot(x,y,color='green',linestyle='dashed',marker='*',markersize=10)
plt.title('this is my first line chart')
plt.xlabel('this is the X axis')
plt.ylabel('this is the Y axis')
plt.legend(['Line'])
plt.grid(True)
plt.show()
```

### 2.Bar Chart

#### A bar plot or bar chart that represents the category of data with length & height that is proportional to the values which they represent, the bar plot can be plotted horizontally or vertically. a bar chart describes the comparison b/w the discrete categories, it can be created by using the bar() method.

```python
x=[3,4,6,7,8,12,15]
y=[4,6,5,8,3,6,1]
plt.bar(x,y,color='orange',width=0.5,edgecolor='black')
plt.plot(x,y,linestyle='dashed',color='green',marker='o',markersize=15)
plt.title('Bar Chart')
plt.legend('Bar')
plt.grid(True)
plt.show()
```

### Histogram
#### it is basically used to represemt data in the form of some groups. its is type of bar chart where the x axis represents the bin range while the y axis gives info abt frequency. to create a histogram the first steo is to create a bin of the ranges, then distrobute the wholw range of the values inot a series of intervals & count thr values which fall into each of the intervals. Bins are clearly identified as consecutive non-overlapping intervals of varialbles.

```python
x=[6,2,2,6,10,45,23,23,34,56,12,34,78,89,3,12]
plt.hist(x,bins=5)
plt.title('Histogram')
plt.legend('Bar')
plt.show()
```

```python
companies = ['Jio', 'Airtel', 'VI', 'BSNL', 'MTNL']
market_share = [40, 35, 18, 65, 42]  # sample percentages
ex=[0,0.3,0,0.2,0]
# Create pie chart
plt.pie(market_share,labels=companies,autopct='%1.3f%%',startangle=120,explode=ex)

# Add title
plt.title("Market Share of Telephone Companies in India")

# Display chart
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
days = [1, 2, 3, 4, 5]
apples = [10, 12, 15, 14, 16]
bananas = [8, 9, 11, 10, 12]
oranges = [5, 6, 7, 8, 9]

# Create stacked chart
plt.stackplot(days, apples, bananas, oranges, labels=['Apples', 'Bananas', 'Oranges'],alpha=0.5,linestyle='dashed',edgecolor='black')

# Add labels and title
plt.xlabel("Day")
plt.ylabel("Quantity Sold")
plt.title("Simple Stacked Area Chart")

# Add legend
plt.legend()
plt.grid(True)
# Show plot
plt.show()
```

## Area Chart

```python
days = [1, 2, 3, 4, 5]
sales = [10, 15, 13, 18, 20]

# Create area chart
plt.fill_between(days, sales,alpha=0.6,linestyle='dashed',edgecolor='black')

# Add labels and title
plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Single Area Chart")

# Show plot
plt.show()
```
