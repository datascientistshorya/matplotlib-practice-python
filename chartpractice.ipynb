```python
import matplotlib.pyplot as plt
```

### 1 Line Chart — Monthly Electricity Consumption (kWh)


```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
electricity_usage = [320, 300, 350, 420, 480, 510]
plt.plot(months,electricity_usage,color='orange',marker='',linewidth=2,markersize=10,linestyle=':')
plt.xlabel("Months")
plt.ylabel("Usage")
plt.title("Monthly Electricity Consumption (kWh)")
plt.grid(True,linestyle='--',color='pink')
plt.legend(['Line'])
plt.show()
```

### Bar Chart — Department-wise Monthly Sales (₹ in Lakhs)

```python
departments = ['Electronics', 'Clothing', 'Grocery', 'Furniture', 'Toys']
sales = [85, 60, 120, 45, 30]
plt.bar(departments,sales,color='red',edgecolor='black',alpha=0.3)
plt.plot(departments,sales,color='orange',linestyle='--',marker='^',markersize=10)
plt.title('Department-wise Monthly Sales (₹ in Lakhs)')
plt.xlabel('Departments')
plt.ylabel('Sales')
plt.grid(True)
plt.legend(['BAR'])
plt.show()
```

## Histogram — Exam Scores of Students

```python
scores = [
    45, 50, 52,58, 60, 62,65,65, 65, 68, 70,
    72, 75, 78, 80, 82, 90, 92, 95
]
b=[40,50,60,70,80,90,100]
plt.hist(scores,bins=b,color='yellow',alpha=0.7,edgecolor='black')
plt.title('Exam Scores of Students')
plt.xlabel('Scores')
plt.ylabel('Frequency')
plt.grid(True,color='pink',linestyle='--')
plt.show()
```

## Pie Chart — Mobile Network Market Share (%)

```python
companies = ['Jio', 'Airtel', 'VI', 'BSNL', 'Others']
market_share = [41, 34, 27, 16, 12]
color=['cyan','yellow','pink','blue','orange']
e=[0.04,0,0.06,0,0]
plt.pie(market_share,labels=companies,autopct='%1.2f%%',startangle=90,explode=e,colors=color,shadow=True)
plt.legend(['pie'])
plt.title(' Mobile Network Market Share (%)')
plt.show()
```

## Scatter Plot — Study Hours vs Exam Marks

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
exam_marks = [40, 45, 55, 60, 68, 75, 82, 90]
plt.scatter(study_hours,exam_marks,color='red',alpha=0.7,marker='v')
plt.grid(True,color='green',linestyle='--')
plt.title('Study Hours vs Exam Marks',color='green')
plt.xlabel('Study Hours',color='green')
plt.ylabel('Marks',color='green')
plt.legend(['v'])
plt.show()
```

## Area Chart — Daily Website Visitors

```python
days = [1, 2, 3, 4, 5, 6, 7]
visitors = [120, 150, 180, 170, 200, 230, 260]
plt.fill_between(days,visitors,alpha=0.3,edgecolor='green',linestyle='--')

plt.title('Daily Website Visitors',color='blue')
plt.xlabel('Days',color='blue')
plt.ylabel('Vistors',color='blue')
plt.show()
```

## Stacked Area Chart — Product-wise Weekly Sales

```python
days = [1, 2, 3, 4, 5]

phones = [20, 25, 30, 35, 40]
laptops = [15, 18, 20, 22, 25]
tablets = [10, 12, 15, 14, 18]
plt.stackplot(days,phones,laptops,tablets,labels=['phones','laptops','tablets'],alpha=0.5,linestyle='dashed',edgecolor='black',colors=['pink','cyan','red'])

plt.xlabel('Products')
plt.ylabel('Sales')
plt.title('Product-wise Weekly Sales')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

data = [12,15,14,10,18,20,22,25,24,30,28,27]

plt.figure(figsize=(8,5))

plt.boxplot(data, notch=True, patch_artist=True,
            flierprops=dict(marker='o',color='yellow', markersize=8),
            showmeans=True)

plt.title("Improved Boxplot Example")
plt.ylabel("Values")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

data = [10,12,14,15,18,20,22,25,28,30]

plt.figure(figsize=(6,4))

plt.boxplot(data, patch_artist=True, vert=False, showmeans=True,
            boxprops=dict(facecolor='lightgreen', color='darkgreen'),
            medianprops=dict(color='red', linewidth=2),
            whiskerprops=dict(color='blue', linewidth=2),
            capprops=dict(color='purple'),
            flierprops=dict(marker='o', markerfacecolor='orange', markersize=8))

plt.title("Colored Boxplot Example")
plt.grid(True,color='red',linestyle='--')

plt.show()
```
